# Data types and structures theory answer :-


1. What are data structures, and why are they important?


 -> Data structures are ways to organize and store data in a computer so it can be used efficiently. Examples include lists, dictionaries, sets, and tuples.

Why they matter:
They help in writing efficient code by allowing fast data access, modification, and organization.

2. Difference between mutable and immutable data types (with examples)


 -> Mutable: Can be changed after creation.
Example: list, dict, set

my_list = [1, 2]
my_list.append(3)  # Modified the list


Immutable: Cannot be changed after creation.
Example: int, str, tuple

name = "Alice"
name[0] = "a"  # ❌ Error: strings are immutable


3. Main differences between lists and tuples in Python

 -> Feature	List ([])	Tuple (())
Mutability	Mutable	Immutable
Speed	Slower	Faster
Use case	Changing data	Fixed data
Syntax	[1, 2, 3]	(1, 2, 3)


4. How do dictionaries store data?

 -> Dictionaries store data as key-value pairs using a structure called a hash table.

Example:

student = {'name': 'Tom', 'age': 20}


'name' is the key → 'Tom' is the value.



5. Why might you use a set instead of a list in Python?

 -> Sets remove duplicates automatically.

They provide faster membership tests (in) than lists.

Example:

my_set = {1, 2, 2, 3}
print(my_set)  # {1, 2, 3}



6. What is a string in Python, and how is it different from a list?

 -> A string is a sequence of characters (e.g., "hello").

A list is a sequence that can hold any data types.

Key difference: Strings are immutable, lists are mutable.



7. How do tuples ensure data integrity in Python?

 -> Tuples are immutable, meaning their contents cannot be changed after creation. This prevents accidental modification, making them ideal for fixed or constant data.




8. What is a hash table, and how does it relate to dictionaries in Python?

 -> A hash table is a data structure that maps keys to values using a hash function.

In Python, dictionaries use hash tables to store and access data quickly.




9. Can lists contain different data types in Python?

 -> Yes. A Python list can hold mixed data types:

my_list = [1, "hello", 3.14, True]




10. Why are strings immutable in Python?

 -> Strings are immutable for:

Security (e.g., used as keys in dictionaries)

Performance (reusability via interning)

Thread safety




11. What advantages do dictionaries offer over lists for certain tasks?

 -> Dictionaries:

Allow quick lookup by keys (O(1) time)

Are better for structured data (like records)

Example:

users = {'id_1': 'Alice', 'id_2': 'Bob'}




12. Scenario where using a tuple is better than a list

 -> Use a tuple when:

The data should not change

You want to use it as a dictionary key

Example:

location = (40.7128, -74.0060)  # Coordinates (latitude, longitude)




13. How do sets handle duplicate values in Python?

 -> Sets automatically remove duplicates.

s = {1, 2, 2, 3}
print(s)  # {1, 2, 3}




14. How does the “in” keyword work differently for lists and dictionaries?

 -> List: Checks if value is in the list → O(n) time

Dict: Checks if key exists → O(1) time

3 in [1, 2, 3]       # True (checks values)
'name' in {'name': 'Tom'}  # True (checks keys)




15. Can you modify elements of a tuple? Why or why not?

 -> No, tuples are immutable. Once created, you cannot change their elements.

t = (1, 2, 3)
t[0] = 5  # ❌ Error

16. What is a nested dictionary? Give an example.

 -> A nested dictionary is a dictionary inside another dictionary.

Example:

students = {
    'John': {'age': 20, 'grade': 'A'},
    'Jane': {'age': 22, 'grade': 'B'}
}




17. Time complexity of accessing elements in a dictionary

 -> Average case: O(1)

Worst case (rare): O(n), due to hash collisions'




18. When are lists preferred over dictionaries?

 -> Use lists when:

Order matters

You only need values, not key-value pairs

You need to maintain duplicates




19. Why are dictionaries considered unordered (before Python 3.7)?

 -> Before Python 3.7, dictionaries didn't preserve insertion order.
Now (Python 3.7+), they do maintain order, but this is still not the main purpose.

Retrieval is by key, not by position.



20. Difference between a list and a dictionary in terms of data retrieval

 ->

| Feature      | List                    | Dictionary                  |
| ------------ | ----------------------- | --------------------------- |
| Retrieval    | By index (0, 1, 2, ...) | By key ('name', 'age', ...) |
| Access speed | O(1) by index           | O(1) by key (on average)    |
| Example      | `my_list[0]`            | `my_dict['name']`           |
